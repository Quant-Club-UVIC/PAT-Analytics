# Strat Comparison

We will write a custom strategy, and then compare it to a couple of built-in strategies.

```python
from pathlib import Path
import pandas as pd
import plotly.graph_objects as go

from pat_analytics import Market, Portfolio
from pat_analytics.strategy import BuyNHold, EqualWeight ,StratConfig
from pat_analytics.backtesters import Backtester
from pat_analytics.metrics import RiskReport, PerformanceReport
```

```python
data_dir = Path.cwd() / "sample-data"

market = Market.from_directory(data_dir, date_col="timestamp")
```

```python
port = Portfolio(market, init_weight='uniform')
```

```python
strat = BuyNHold()
```

```python
bt = Backtester(port, market, strat)

bt.run()
```

```python
pr = PerformanceReport(bt.portfolio, market)
pr.plot_returns(by_constituent=True).show()
```

```python
pr.total_return()
```

```python
strat = EqualWeight(config=StratConfig(commission_rate=0.001))
bt = Backtester(port, market, strat)
bt.run()
```

```python
pr = PerformanceReport(bt.portfolio, market)
pr.plot_returns(by_constituent=False).show()
```
